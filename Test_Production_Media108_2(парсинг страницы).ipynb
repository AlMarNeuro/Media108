```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, parse_qs

# Замените 'YOUR_FOLDER_URL' на URL вашей папки Google Диск
folder_url = "https://drive.google.com/drive/folders/1Wo4RMSeEaFeDFU-puHlB650GXQ2gayxF"

def get_drive_links(folder_url):
    response = requests.get(folder_url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Ищем все ссылки в HTML-коде страницы
    #links = soup.find_all('a')
    links = soup.find_all()

    #print(links)

    # Фильтруем только те ссылки, которые ведут на файлы
    file_links = [link['href'] for link in links if 'file/d/' in link['href']]

    print(file_links)

    # Извлекаем идентификаторы файлов из URL
    file_ids = [parse_qs(urlparse(link).query)['id'][0] for link in file_links]

    # Создаем прямые ссылки на скачивание файлов
    download_links = [f'https://drive.google.com/uc?id={file_id}' for file_id in file_ids]

    return download_links

# Получаем список прямых ссылок на скачивание файлов из папки
file_links = get_drive_links(folder_url)

# Выводим список ссылок
for link in file_links:
    print(link)
```
